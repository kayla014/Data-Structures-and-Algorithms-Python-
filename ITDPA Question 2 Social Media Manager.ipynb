```python
# Import the necessary libraries
import networkx as nx
import matplotlib.pyplot as plt

# Connections Manager class
class ConnectionsManager:
    def __init__(self):
        # Store all users
        self.users = set()
        # Store user connections
        self.connections = {}
        # Graph object for visualization
        self.graph = nx.Graph()

    # Add a user
    def add_user(self, user):
        if user in self.users:
            print(f"User '{user}' already exists.")
        else:
            self.users.add(user)
            self.connections[user] = []
            self.graph.add_node(user)
            print(f"User '{user}' has been added successfully.")

    # Add a connection between two users
    def add_connection(self, user1, user2):
        if user1 not in self.users or user2 not in self.users:
            print("Both users must exist to create a connection.")
            return
        if user2 in self.connections[user1]:
            print(f"Connection between '{user1}' and '{user2}' already exists.")
            return
        self.connections[user1].append(user2)
        self.connections[user2].append(user1)
        self.graph.add_edge(user1, user2)
        print(f"Connection between '{user1}' and '{user2}' has been added successfully.")

    # View all users
    def view_users(self):
        if not self.users:
            print("No users in the network.")
        else:
            print("\nUsers in the network:")
            for user in sorted(self.users):
                print("-", user)

    # View all connections
    def view_connections(self):
        if not self.connections:
            print("No connections in the network.")
        else:
            print("\nConnections in the network:")
            for user, friends in self.connections.items():
                print(f"{user}: {', '.join(friends) if friends else 'No connections'}")

    # Display the network graph
    def display_graph(self):
        if not self.users:
            print("No users to display in the graph.")
            return
        plt.figure(figsize=(8, 6))
        nx.draw(
            self.graph,
            with_labels=True,
            node_color="lightblue",
            edge_color="gray",
            node_size=2000,
            font_size=12
        )
        plt.title("Social Media Manager")
        plt.show()


# Main menu loop
def main():
    manager = ConnectionsManager()

    while True:
        print("\nSocial Media Connections Manager")
        print("1. Add a user")
        print("2. Add a connection")
        print("3. View all users")
        print("4. View all connections")
        print("5. Display the network graph")
        print("6. Quit")

        choice = input("Please enter your choice: ")

        if choice == "1":
            user = input("Please enter the user name: ")
            manager.add_user(user)

        elif choice == "2":
            user1 = input("Please enter the first user: ")
            user2 = input("Please enter the second user: ")
            manager.add_connection(user1, user2)

        elif choice == "3":
            manager.view_users()

        elif choice == "4":
            manager.view_connections()

        elif choice == "5":
            manager.display_graph()

        elif choice == "6":
            print("Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
