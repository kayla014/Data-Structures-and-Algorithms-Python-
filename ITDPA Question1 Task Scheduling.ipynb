```python
import heapq
from datetime import datetime

# Task class
class Task:
    def __init__(self, description, due_date, priority, estimated_time):
        self.description = description
        self.due_date = due_date
        self.priority = priority
        self.estimated_time = estimated_time

    def __lt__(self, other):
        return self.priority < other.priority

    # Display task info
    def print_task_info(self):
        print("Task:", self.description)
        print("Due date:", self.due_date.strftime('%Y-%m-%d'))
        print("Priority:", self.priority)
        print("Estimated time to complete:", self.estimated_time, "hrs")
        print("-" * 40)

    # Converts task to string for file saving
    def to_string(self):
        return f"{self.description}|{self.due_date.strftime('%Y-%m-%d')}|{self.priority}|{self.estimated_time}"


#  Creates a task from a string
def task_from_string(task_str):
    description, due_date_str, priority, estimated_time = task_str.strip().split("|")
    due_date = datetime.strptime(due_date_str, "%Y-%m-%d")
    return Task(description, due_date, int(priority), float(estimated_time))


# Scheduler class
class Scheduler:
    def __init__(self):
        self.tasks = []
        heapq.heapify(self.tasks)

    # Add task to scheduler
    def add_task(self, task):
        heapq.heappush(self.tasks, task)
        print("Task has been added successfully.")

    # Get next task by priority
    def get_next_task(self):
        if not self.tasks:
            print("No tasks available.")
            return None
        task = heapq.heappop(self.tasks)
        print("Next task to execute:")
        task.print_task_info()
        return task

    # Print all tasks
    def print_all_tasks(self):
        if not self.tasks:
            print("No tasks available.")
            return
        for task in sorted(self.tasks):
            task.print_task_info()

    # Save tasks to file
    def save_tasks_to_file(self, filename):
        with open(filename, "w") as file:
            for task in sorted(self.tasks):
                file.write(task.to_string() + "\n")
        print("Tasks have been saved successfully.")

    # Load tasks from file
    def load_tasks_from_file(self, filename):
        try:
            with open(filename, "r") as file:
                self.tasks = [task_from_string(line) for line in file]
            heapq.heapify(self.tasks)
            print("Tasks have been loaded successfully.")
        except FileNotFoundError:
            print("File has not been found.")


# Main menu loop
def main():
    scheduler = Scheduler()

    while True:
        print("\nTask scheduler menu")
        print("1. Add new task")
        print("2. Get next task")
        print("3. Print all tasks")
        print("4. Save tasks to file")
        print("5. Load tasks from file")
        print("6. Exit")

        choice = input("Please enter your choice: ")

        if choice == "1":
            description = input("Please enter the task description: ")
            due_date_str = input("Please enter the due date (yyyy-mm-dd): ")
            due_date = datetime.strptime(due_date_str, "%Y-%m-%d")
            priority = int(input("Please enter the priority (lower number = higher priority): "))
            estimated_time = float(input("Please enter the estimated time to complete (hours): "))
            task = Task(description, due_date, priority, estimated_time)
            scheduler.add_task(task)

        elif choice == "2":
            scheduler.get_next_task()

        elif choice == "3":
            scheduler.print_all_tasks()

        elif choice == "4":
            filename = input("Please enter the filename to save: ")
            scheduler.save_tasks_to_file(filename)

        elif choice == "5":
            filename = input("Please enter the filename to load: ")
            scheduler.load_tasks_from_file(filename)

        elif choice == "6":
            print("Exiting program.")
            break

        else:
            print("Invalid option. Please try again.")


if __name__ == "__main__":
    main()
```
